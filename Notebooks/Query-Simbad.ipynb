```python
from astroquery.simbad import Simbad
import pandas as pd
from astroquery.utils.tap.core import Tap
```

```python
# Usar el servicio TAP directamente
tap = Tap(url="http://simbad.u-strasbg.fr/simbad/sim-tap")
```

```python
query = """
SELECT main_id, ra, dec, otype, flux 
FROM basic 
JOIN fluxes ON basic.oid = fluxes.oidref 
WHERE 
    1=CONTAINS(OType(), 'YSO') AND
    fluxes.filter = 'G_BPmag' AND
    fluxes.flux > 15
"""
```

```python
job = tap.launch_job(query)
yso_table = job.to_table().to_pandas()
```
